# 01 - Financial News Data Exploration
Explore the Financial PhraseBank dataset, data quality, class distribution, text length, and common financial terms.

```python
import pandas as pd
import matplotlib.pyplot as plt
from src.data_loader import load_financial_phrasebank
from src.feature_engineering import add_text_features
df = add_text_features(load_financial_phrasebank())
df.head()
```

```python
print('Shape:', df.shape)
print('Missing values:\n', df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
df.info()
```

```python
print(df['sentiment'].value_counts())
df['sentiment'].value_counts().plot(kind='bar', title='Financial News Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Number of Articles')
plt.tight_layout()
plt.show()
```

```python
df[['word_count','char_count']].describe()
df['word_count'].plot(kind='hist', bins=30, title='Distribution of News Text Length')
plt.xlabel('Word Count')
plt.tight_layout()
plt.show()
```

```python
from collections import Counter
import re
stop_words = {'the','and','to','of','in','a','for','on','is','that','with','as','by','at','from','it','this','be','are','was','an','or','its','has','have'}
tokens = re.findall(r'\b[a-zA-Z]{3,}\b', ' '.join(df['text'].astype(str)).lower())
counts = Counter(t for t in tokens if t not in stop_words)
common = pd.DataFrame(counts.most_common(20), columns=['term','frequency'])
common
```

```python
common.set_index('term')['frequency'].sort_values().plot(kind='barh', title='Most Frequent Financial Terms')
plt.xlabel('Frequency')
plt.tight_layout()
plt.show()
```
