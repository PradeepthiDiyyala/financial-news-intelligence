# 02 - Data Preprocessing
Build a reproducible preprocessing pipeline for financial news text and inspect the cleaned data.

```python
import pandas as pd
from src.data_loader import load_financial_phrasebank
from src.preprocessing import preprocess_dataframe
df_raw = load_financial_phrasebank()
print('Raw shape:', df_raw.shape)
df_raw.head()
```

```python
df = preprocess_dataframe(df_raw.copy())
print('Processed shape:', df.shape)
df.head()
```

```python
print('Missing values after preprocessing:\n', df.isna().sum())
print('Duplicate rows after preprocessing:', df.duplicated().sum())
print(df['sentiment'].value_counts())
```

```python
comparison = pd.DataFrame({
    'raw_text': df_raw['text'].astype(str),
    'processed_text': df['text'].astype(str)
})
comparison.head(10)
```

```python
df['processed_word_count'] = df['text'].str.split().str.len()
df[['processed_word_count']].describe()
```
