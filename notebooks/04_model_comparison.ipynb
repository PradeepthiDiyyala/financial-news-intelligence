# 04 - Model Comparison
Train the TF-IDF + Logistic Regression baseline, inspect classification metrics, and compare results across sentiment classes.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report
from src.data_loader import load_financial_phrasebank
from src.train_models import build_model
df = load_financial_phrasebank()
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.20, stratify=df['label'], random_state=42)
```

```python
model = build_model()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print(classification_report(y_test, predictions, target_names=['negative','neutral','positive'], digits=4))
```

```python
precision, recall, f1, _ = precision_recall_fscore_support(y_test, predictions, average='macro')
accuracy = accuracy_score(y_test, predictions)
metrics = pd.DataFrame({
    'Metric': ['Accuracy','Macro Precision','Macro Recall','Macro F1'],
    'Score': [accuracy, precision, recall, f1]
})
metrics
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix(y_test, predictions)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['negative','neutral','positive'])
disp.plot()
disp.ax_.set_title('Sentiment Classification Confusion Matrix')
disp.figure_.tight_layout()
disp.figure_.show()
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(build_model(), df['text'], df['label'], cv=5, scoring='f1_macro')
print('5-fold Macro F1 scores:', scores)
print('Mean Macro F1:', scores.mean())
print('Std Macro F1:', scores.std())
```

```python
results = pd.DataFrame({
    'Model': ['TF-IDF + Logistic Regression'],
    'Accuracy': [accuracy],
    'Macro F1': [f1]
})
results
```
